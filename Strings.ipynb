```python
#program that count the no of vowels and consonents in a string
s=str(input("enter a string:"))
b=s.lower()
vowels="aeiou"
count_v=0
count_c=0
for i in b:
  if i.isalpha():
    if i in vowels:
      count_v+=1
    else:
      count_c+=1
print(count_v)
print(count_c)


```

```python
#program that checks,the sub string is in the original string if yes print index of substring
s=str(input("enter a string:"))
b=str(input("enter sub string:"))
if b in s:
  print(s.find(b))
else:
  print("not present")
```

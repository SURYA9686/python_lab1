**TUPLE OPERATIONS**

```python
t1=(1,2,3)
t2=(4,5,6)
print(t1+t2)
print(t1*4)
for i in t1:
  print(i)
print(len(t1))
if "h" in t1:
  print("yes")
else:
  print("no")
```

**functions using tuples**

```python
print(sum(t1))
print(min(t2))
```

**METHODS USING TUPLE**

```python
print(t1.count(2))
print(t2.index(4))
```

**CREATING TUPLE**



```python
t=tuple(map(int,input().split()))
print(t)
```

```python
t=tuple(int(i) for i in input().split())
print(t)
```

ZIPPING

```python
print(tuple(zip(t1,t2)))
```

**TUPLE INSIDE LIST OR LIST INSIDE TUPLE**

```python
t4=(1,2,3,(4,5,6))
print(t4)
```

```python
t5=(1,2,3,[4,5,6])
print(t5)
```

```python
print(t4[3][2])
```

```python
t5[3][1]="hello"
print(t5)
```

write a program to count the number of vowels in a string (no control flow allowed)


```python
s=str(input("enter a string:"))
s1=s.lower()
vowels_count=s1.count("a")+s1.count("e")+s1.count("i")+s1.count("o")+s1.count("u")
print(vowels_count)
```

```python

# Creating a file and writing data

f = open("demo.txt", "w")
f.write("Hello Python\n")
f.write("Welcome to File Handling")
f.close()

print("File created successfully")
```

```python

# Reading a file

f = open("demo.txt", "r")
print(f.read())
f.close()
```

```python

# Adding data to an existing file

f = open("demo.txt", "a")
f.write("\nThis is appended text")
f.close()

print("Data appended successfully")
```

```python

f = open("demo.txt", "r")
print(f.read(5))
f.close()
```

```python

f = open("demo.txt", "r")
print(f.readline())
f.close()
```

```python

f = open("demo.txt", "r")

for i in f:
    print(i)

f.close()
```

```python

with open("demo.txt", "r") as f:
    print(f.read())
```

```python

with open("demo.txt", "r") as f:
    data = f.read()
    print("Number of characters:", len(data))
```
